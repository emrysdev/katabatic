```python
import os
import pandas as pd
from scipy.io import arff
import yaml

# Your input and output directories
input_dir = r"C:\Deakin\Project\katabatic\tabddpm\TabDDPM\tab-ddpm\tab_ddpm\arff\input_files"
output_dir = r"C:\Deakin\Project\katabatic\tabddpm\TabDDPM\tab-ddpm\tab_ddpm\arff\output_files"
os.makedirs(output_dir, exist_ok=True)

# List of ARFF filenames in the input directory
arff_files = [
    "adult.arff",
    "chess.arff",
    "letter-recog.arff",
    "connect-4.arff",
    "car 1.arff",
    "nursery.arff",
    "credit-a.arff",
    "dermatology.arff"
]

# Convert each file
for file_name in arff_files:
    input_path = os.path.join(input_dir, file_name)
    name = os.path.splitext(file_name)[0].replace(" ", "_").replace("(", "").replace(")", "")
    
    # Load ARFF and convert to DataFrame
    data, _ = arff.loadarff(input_path)
    df = pd.DataFrame(data)

    # Save as CSV
    csv_path = os.path.join(output_dir, f"{name}.csv")
    df.to_csv(csv_path, index=False)

    # Save metadata as YAML
    yaml_data = {
        "name": name,
        "path": csv_path,
        "features": df.columns.tolist(),
        "num_features": len(df.columns),
        "num_instances": len(df)
    }
    yaml_path = os.path.join(output_dir, f"{name}.yaml")
    with open(yaml_path, "w") as f:
        yaml.dump(yaml_data, f)

    print(f"✅ Converted: {file_name} → {csv_path}, {yaml_path}")
```
